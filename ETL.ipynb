## Ocupacao Covid

Para os valores de 2020 e 2021, as ocupações dos leitos eram divididas em ocupação por suspeita e ocupação quando o diagnóstico de COVID era confirmado, isso levou ao grande número de NaNs nas colunas referentes às ocupações dos leitos, para resolver isso os campos NaN de ocupacaoCovidUti e ocupacaoCovidCli somando ocupacaoSuspeito e ocupacaoConfirmado para Cli ou UTI correspondentemente

```python
mask = (
    df['ano_origem'].isin(['2020', '2021']) &
    df['ocupacaoCovidUti'].isna()
)

df.loc[mask, 'ocupacaoCovidUti'] = (
    df.loc[mask, 'ocupacaoSuspeitoUti'].fillna(0) +
    df.loc[mask, 'ocupacaoConfirmadoUti'].fillna(0)
)

mask = (
    df['ano_origem'].isin(['2020', '2021']) &
    df['ocupacaoCovidCli'].isna()
)

df.loc[mask, 'ocupacaoCovidCli'] = (
    df.loc[mask, 'ocupacaoSuspeitoCli'].fillna(0) +
    df.loc[mask, 'ocupacaoConfirmadoCli'].fillna(0)
)
```

# Tipo de dados

Alterar o tipo de dataNotificao,created_at e updated_at para datetime64, assim essa coluna poderá ser usada para gerar Dia, Mês e Hora.

```python
df['dataNotificacao'] = pd.to_datetime(df['dataNotificacao'])
df['dataCriado'] = pd.to_datetime(df['_created_at'])
df['dataAtualizado'] = pd.to_datetime(df['_updated_at'])
```

```python
colunas_hospital = ['cnes','estado','municipio']

dim_hospital = df[colunas_hospital].drop_duplicates().reset_index(drop=True)
dim_hospital = dim_hospital.reset_index().rename(columns={'index': 'id_hospital_sk'})

print(f"Dimensão Hospital: {len(dim_hospital)} perfis")

colunas_notificacao = ['origem','_p_usuario','estadoNotificacao','municipioNotificacao']

dim_notificacao = df[colunas_notificacao].drop_duplicates().reset_index(drop=True)
dim_notificacao = dim_notificacao.reset_index().rename(columns={'index': 'id_notificacao_sk'})

print(f"Dimensão Notificação: {len(dim_notificacao)} perfis")

colunas_tempo = ['ano_origem','dia_notificacao','mes_notificacao','hora_notificao','dia_criado','mes_criado','hora_criado','dia_atualizado','mes_atualizado','hora_atualizado']

dim_tempo = df[colunas_tempo].drop_duplicates().reset_index(drop=True)
dim_tempo = dim_tempo.reset_index().rename(columns={'index': 'id_tempo_sk'})

print(f"Dimensão Tempo: {len(dim_tempo)} perfis")
```

```python
fato = df.merge(dim_hospital, on=colunas_hospital, how='left')
fato = fato.merge(dim_notificacao, on=colunas_notificacao, how='left')
fato = fato.merge(dim_tempo, on=colunas_tempo, how='left')

#Pegando apenas os IDs e a dose
colunas_fato = [
    'id_hospital_sk', 
    'id_notificacao_sk',
    'id_tempo_sk', 
    'ocupacaoCovidUti',
    'ocupacaoCovidCli',
    'saidaSuspeitaObitos',
    'saidaSuspeitaAltas',
    'saidaConfirmadaObitos',
    'saidaConfirmadaAltas'
]

fato_covid = fato[colunas_fato]
print(f"Fato pronta: {len(fato_covid)} registros.")
```

```python
from sqlalchemy import create_engine

# Configurando a conexão
usuario_db = ''               
senha_db = ''                
host_db = ''                    
porta_db = ''                    
nome_banco_db = ''          

# Criando a "connection string"
DATABASE_URL = f'postgresql://{usuario_db}:{senha_db}@{host_db}:{porta_db}/{nome_banco_db}?client_encoding=utf8'
print("Tentando criar conexão com o banco...")

try:
    engine = create_engine(DATABASE_URL)
    print("Conexão bem-sucedida! Iniciando a carga do Esquema Estrela...")
    
    print("Carregando Dimensão Paciente...")
    dim_hospital.to_sql('dim_hospital', con=engine, if_exists='replace', index=False)
    
    print("Carregando Dimensão Localidade...")
    dim_notificacao.to_sql('dim_hospital', con=engine, if_exists='replace', index=False)
    
    print("Carregando Dimensão Vacina...")
    dim_tempo.to_sql('dim_tempo', con=engine, if_exists='replace', index=False)
    
   

    
    print("Carregando Tabela Fato...")
    fato_covid.to_sql('fato_covid', con=engine, if_exists='replace', index=False)

    print("\nCARGA CONCLUÍDA! Todas as 6 tabelas do Esquema Estrela foram criadas.")

except Exception as e:
    print(f"\n--- ERRO! ---")
    print(f"Erro ao carregar os dados: {e}")
```
